## Step 1: payoffs

With $k$ contributors, a contributor gets $R + s + r - (c - f) + n_b k + B[k \ge k^*]$ and a harvester $R + n_b k + B[k \ge k^*]$: $s$ private benefit, $r$ recognition, $c$ cost, $f$ funding, $n_b$ what each contribution gives everyone, and $B$ what a maintained commons gives everyone once $k^*$ contributors sustain it.

```python
#@title Your game { display-mode: "form" }
from hecon import game
players = 8  #@param {type:"integer"}
s = 2.0  #@param {type:"number"}
r = 0.0  #@param {type:"number"}
c = 6.0  #@param {type:"number"}
f = 0.0  #@param {type:"number"}
nb = 1.0  #@param {type:"number"}
B = 8.0  #@param {type:"number"}
k_star = 3  #@param {type:"integer"}
G = dict(players=players, s=s, r=r, c=c, f=f, nb=nb, B=B, k_star=k_star)
eq = game.equilibria(G); opt = game.optimum(G)
print("Equilibria (number of contributors):", eq)
print("Social optimum:", opt, "contributors")
for k in sorted(set(eq + [opt])): print("  welfare with %d contributors: %.1f" % (k, game.welfare(k, G)))
```

## Step 2: reading it

With the defaults there are two equilibria, nobody contributes or exactly $k^*$ do, and both fall short of the optimum. That is the collective-action problem of section 9.3: cultivation can be stable, but only once enough others cultivate.

## Step 3: how a population drifts

If institutions copy what pays, the share that cultivates follows the replicator dynamic below. Starting points above the tipping share climb; those below fall away.

```python
#@title Replicator dynamics { display-mode: "form" }
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 4))
for x0 in [0.05, 0.15, 0.25, 0.35, 0.5, 0.7, 0.9]:
    plt.plot(game.replicator(x0, G, steps=120), label="start %.2f" % x0)
plt.ylim(0, 1); plt.xlabel("rounds"); plt.ylabel("share cultivating"); plt.legend(fontsize=7); plt.show()
```

## Step 4: what moves the equilibrium

Recognition $r$ and funding $f$ both change the private return to contributing. The table shows the equilibria for each pair.

```python
#@title Recognition and funding { display-mode: "form" }
for rr in [0, 1, 2, 3, 4]:
    print("r=%d " % rr + "  ".join("f=%d:%s" % (ff, game.equilibria(dict(G, r=rr, f=ff))) for ff in [0, 1, 2, 3]))
```
